```python
import pandas as pd
import duckdb
import matplotlib.pyplot as plt
import seaborn as sns
from adjustText import adjust_text 

df = pd.read_csv("../data/players_data_light-2025_2026.csv").convert_dtypes()


df.head()
```

```python
# Ta bara med spelare som har spelat minst 5 hela matcher (450 minuter totalt)
# Detta görs för att få bort sena inhoppare som kan förvärra statistiken
cleaned_df = df[df['90s'] >= 5.0].copy()

# Räknar hur många spelare som är kvar efter rensningen samt skriver ut resultatet efter rensning
liga_query = """
    SELECT Comp, COUNT(*) as AntalSpelare
    FROM cleaned_df
    GROUP BY Comp
    ORDER BY AntalSpelare DESC
"""
liga_stats = duckdb.query(liga_query).to_df()


print(f"Antal spelare från början: {len(df)}")
print(f"Antal spelare efter rensning: {len(cleaned_df)}\n")
print("Antal rensade spelare per liga:")
print(liga_stats)


cleaned_df.head()
```

```python
# Vilka 5 spelare har gjort flest mål (Gls)?
top_scorers_query = """
    SELECT Player, Squad, Comp, Gls
    FROM cleaned_df
    ORDER BY Gls DESC
    LIMIT 5
"""

top_scorers_df = duckdb.query(top_scorers_query).to_df()
top_scorers_df
```

```python
# Vilka är de 5 mest effektiva unga spelarna (under 23 år) baserat på mål + assist (G+A)?
young_stars_query = """
    SELECT Player, Squad, Age, "G+A"
    FROM cleaned_df
    WHERE Age <= 23
    ORDER BY "G+A" DESC
    LIMIT 5
"""
young_stars_df = duckdb.query(young_stars_query).to_df()
young_stars_df
```

```python
# Vilken liga (Comp) är tuffast?
liga_cards_query = """
    SELECT Comp as Liga, 
           COUNT(*) as Antal_Spelare,
           ROUND(AVG(CrdY), 2) as Gula_Kort_Snitt,
           SUM(CrdR) as Totala_Röda_Kort
    FROM cleaned_df
    GROUP BY Comp
    ORDER BY Gula_Kort_Snitt DESC
"""

liga_cards_df = duckdb.query(liga_cards_query).to_df()

liga_cards_df
```

```python
# Djupare analys av skotteffektivitet för spelare med minst 10 mål
efficiency_query = """
    SELECT Player as Spelare, 
           Squad as Lag, 
           Comp as Liga, 
           Gls as Mål, 
           Sh as Skott_Totalt, 
           SoT as Skott_På_Mål,
           ROUND(CAST(Gls AS FLOAT) / CAST(Sh AS FLOAT), 2) as Mål_Per_Skott,
           ROUND(CAST(SoT AS FLOAT) / CAST(Sh AS FLOAT) * 100, 1) as Pricksäkerhet_Procent
    FROM cleaned_df
    WHERE Gls >= 10 AND Sh > 0
    ORDER BY Mål_Per_Skott DESC
    LIMIT 10
"""

efficiency_df = duckdb.query(efficiency_query).to_df()
efficiency_df
```

```python
# Vilka 10 spelare blir fällda (fouled) flest gånger?
most_fouled_query = """
    SELECT Player as Spelare, Squad as Lag, Comp as Liga, Pos as Position, Fld as Fouls_Emot_Sig
    FROM cleaned_df
    ORDER BY Fld DESC
    LIMIT 10
"""

most_fouled_df = duckdb.query(most_fouled_query).to_df()
most_fouled_df
```

```python
# Vilka spelare skjuter mest per 90 minuter?
most_shots_query = """
    SELECT Player as Spelare, Squad as Lag, Comp as Liga, "Sh/90" as Skott_Per_90min, Gls as Totala_Mål
    FROM cleaned_df
    ORDER BY "Sh/90" DESC
    LIMIT 10
"""

most_shots_df = duckdb.query(most_shots_query).to_df()
most_shots_df
```

```python
# Vilka spelare har fått flest röda kort och hur många matcher har de spelat?
most_reds_query = """
    SELECT Player as Spelare, 
           Squad as Lag, 
           Comp as Liga, 
           MP as Matcher,
           CrdR as Röda_Kort, 
           CrdY as Gula_Kort
    FROM cleaned_df
    WHERE CrdR > 0
    ORDER BY CrdR DESC, CrdY DESC, MP ASC
    LIMIT 10
"""


most_reds_df = duckdb.query(most_reds_query).to_df()
most_reds_df
```

```python
# Vilka försvarare gör flest mål?
scoring_defenders_query = """
    SELECT Player as Spelare, Squad as Lag, Comp as Liga, Pos as Position, Gls as Mål, MP as Matcher
    FROM cleaned_df
    WHERE Pos LIKE '%DF%'
    ORDER BY Gls DESC
    LIMIT 10
"""

scoring_defenders_df = duckdb.query(scoring_defenders_query).to_df()
scoring_defenders_df
```

```python
# Stapeldiagramm för bästa målskyttar i Europa
plt.figure(figsize=(10, 5))
sns.barplot(x="Gls", y="Player", data=top_scorers_df, hue="Comp", palette="viridis")


plt.title("Topp 10 Målskyttar i Europa", fontsize=14, pad=15)
plt.xlabel("Antal Mål")
plt.ylabel("Spelare")

plt.tight_layout()
plt.show()
```

```python
most_shots_df = duckdb.sql(most_shots_query).df()

# Sätt storleken på diagrammet och skapa ett liggande stapeldiagram
plt.figure(figsize=(10, 5))
ax = sns.barplot(x="Skott_Per_90min", y="Spelare", data=most_shots_df, color="#2b5c8f")

# Den här loopen lägger till de exakta siffrorna (med två decimaler) längst ut på varje stapel
for p in ax.patches:
    width = p.get_width()
    ax.text(width + 0.05, p.get_y() + p.get_height()/2, f'{width:.2f}', 
            va='center', ha='left', fontsize=10, fontweight='bold')

# Titlar och etiketter för axlarna
plt.title("Topp 10 Spelare med flest skott per 90 minuter", fontsize=14, pad=15, fontweight='bold')
plt.xlabel("Skott per 90 minuter", fontsize=11)
plt.ylabel("Spelare", fontsize=11)


sns.despine()

# Sätter maxgränsen på X-axeln så att siffrorna längst ut till höger får plats ordentligt
plt.xlim(0, 5.5)

plt.tight_layout()
plt.show()
```

```python
# Sortera fram de 30 spelarna som gjort absolut flest mål i Europa
top_30_scorers = cleaned_df.sort_values(by="Gls", ascending=False).head(30).copy()

# Se till att siffrorna tolkas rätt av Python så att diagrammet ritas upp korrekt
top_30_scorers['Fld'] = pd.to_numeric(top_30_scorers['Fld'], errors='coerce')
top_30_scorers['Gls'] = pd.to_numeric(top_30_scorers['Gls'], errors='coerce')

plt.figure(figsize=(13, 8))
sns.scatterplot(x="Fld", y="Gls", hue="Comp", data=top_30_scorers, s=100, palette="Set1")

# Den här loopen går igenom de 30 spelarna och skriver ut deras namn precis bredvid deras prick
for index, row in top_30_scorers.iterrows():
    plt.text(
        x=row['Fld'] + 0.5,       # Skjut texten lite till höger om pricken
        y=row['Gls'] + 0.1,       # Skjut texten lite uppåt
        s=row['Player'],          # Skriv ut namnet
        fontsize=9,
        weight='semibold',
        color='#333333'
    )

plt.title("Topp 30 Målskyttar i Europa: Mål vs Hur ofta de blir fällda", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Antal gånger fälld under säsongen (Fld)", fontsize=11)
plt.ylabel("Totalt antal mål (Gls)", fontsize=11)
plt.legend(title="Liga", bbox_to_anchor=(1.05, 1), loc='upper left') # Lägg ligaförklaringen utanför rutan

plt.xlim(top_30_scorers['Fld'].min() - 2, top_30_scorers['Fld'].max() + 8)
plt.ylim(top_30_scorers['Gls'].min() - 1, top_30_scorers['Gls'].max() + 2)

sns.despine()
plt.tight_layout()
plt.show()
```
